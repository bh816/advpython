## Decorators

#####  1.Input of decorator is a function
#####  2. Output of Decorator is a modified input function (Wrapper)

```python
def hypotenuse(a,b):
    h = (a**2 +b**2)**(1/2)
    return h
```

```python
hypotenuse(3,4)
```

```python
hypotenuse(a=12,b=13)
```

## Welcome Decorator

1.Welcome the User    
2.Calculate results   
3.Thank The User  

```python
def welcome(func):

   def wrapper(*args, **kwargs):
      print("Welcome User")
      result = func(*args, *kwargs)
      print(f"Results : {result}")
      print("Thank you")
      return result 
   
   return wrapper
```

```python
@welcome
def hyp_dec(a,b):
    h = (a**2 +b**2)**(1/2)
    return h
```

```python
h1 = hyp_dec(3,4)
```

```python
h1
```

```python
h2 = hyp_dec(5,6)
```

```python
h2
```

```python
@welcome
def simple_intrest(p, n, r):
    i = (p*n*r)/100
    a = p + i
    return i, a
```

```python
r1 = simple_intrest(p=53000, n=5, r=7.1)
```

### Calculating Function Time With Decorators 

```python
import time
```

```python
print("Hello")
```

```python
time.sleep(2) #python will remain idle for 2 sec
print("Hello")
```

```python
start = time.perf_counter()
time.sleep(5)
print("Hello!")
stop = time.perf_counter()
elapsed = stop - start
print(f"Time Elapsed : {elapsed:.2f} sec")
```

```python
def time_dec(func):

    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        results = func(*args, **kwargs)
        stop = time.perf_counter()
        elapsed = stop - start 
        print(f"Time Elapsed : {elapsed:.2f} sec")
        return results
    
    return wrapper
```

```python
@time_dec
def compound_intrest(p, n, r):
    time.sleep(3.5)
    a = p*(1 + (r/100))**n
    i = a - p
    return i, a
```

```python
r1 = compound_intrest(p=40000, n=3, r=6.5)
```

```python
r1
```

```python
@time_dec
def avg(nums: list) -> int|float:
    # Intialize sum to 0
    s = 0

    # Applied for loop on nums 
    for i in nums:
        time.sleep(2)
        s = s + i
    
    # Calculate number of elements
    c = len(nums)

    return s/c   
```
